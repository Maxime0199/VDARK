## Check mapping quality

```r
get_mapq_at_positions <- function(bam_file, snv_df,
                                   chrom_col = "#CHROM", pos_col = "POS") {
    gr <- GRanges(seqnames = snv_df[[chrom_col]],
                 ranges   = IRanges(start = snv_df[[pos_col]], width = 1))
    param <- ScanBamParam(which = gr, what = c("pos", "mapq", "cigar"))
    res   <- scanBam(bam_file, param = param)   # 1 élément de liste par ligne de snv_df, même ordre
    summarised <- rbindlist(lapply(res, function(r) {
        m <- r$mapq
        if (length(m) == 0) {
            data.table(n_reads = 0L, mean_mapq = NA_real_,
                      min_mapq = NA_real_, frac_mapq0 = NA_real_)
        } else {
            data.table(n_reads    = length(m),
                      mean_mapq  = mean(m, na.rm = TRUE),
                      min_mapq   = min(m, na.rm = TRUE),
                      frac_mapq0 = mean(m == 0, na.rm = TRUE))
        }
    }))
    cbind(snv_df, summarised)
}

get_pileup_metrics_at_positions <- function(bam_file, snv_df,
                                             chrom_col = "#CHROM",
                                             pos_col   = "POS",
                                             alt_col   = "ALT",
                                             min_base_quality = 20,
                                             min_mapq = 0,
                                             by_strand = FALSE) {

    gr <- GRanges(seqnames = snv_df[[chrom_col]],
                  ranges   = IRanges(start = snv_df[[pos_col]], width = 1))

    pparam <- PileupParam(distinguish_strands     = by_strand,
                          distinguish_nucleotides = TRUE,
                          min_base_quality        = min_base_quality,
                          min_mapq                = min_mapq,
                          include_deletions       = FALSE,
                          include_insertions      = FALSE)

    sbparam <- ScanBamParam(which = gr)
    pu <- as.data.table(pileup(bam_file, scanBamParam = sbparam,
                               pileupParam = pparam))

    metrics <- rbindlist(lapply(seq_len(nrow(snv_df)), function(i) {
        chrom <- snv_df[[chrom_col]][i]
        pos_i <- snv_df[[pos_col]][i]
        alt   <- snv_df[[alt_col]][i]

        rows  <- pu[seqnames == chrom & pos == pos_i]
        depth <- sum(rows$count)
        n_alt <- sum(rows$count[rows$nucleotide == alt])

        data.table(depth = depth, n_alt_bam = n_alt, n_ref_bam = depth - n_alt,
                  vaf   = if (depth > 0) n_alt / depth else NA_real_)
    }))

    cbind(snv_df, metrics)
}
```

```r
compare_var <- function(varname, df1, label1,  df2, label2) {
    cat("--------------------------------------------------------\n")
    cat(varname, "\n")
    cat(label1,":\n");  print(summary(df1[[varname]]))
    cat(label2,":\n"); print(summary(df2[[varname]]))
}

vars <- c(
    "n_sig_kmers_in_normal", "aln_score", "n_kmer_tumour_sig",
    "n_tumour_reads", "n_normal_reads", "n_normal_kmers_total",
    "n_mismatches_total", "n_mismatches_somatic", "mean_weight",
    "length_contig_tumour", "length_contig_normal", "mean_mapq",
    "depth", "n_alt", "n_ref", "vaf", "true_mutation"
)

invisible(lapply(vars, compare_var, fp_true, "FP_true", tp_true, "TP_true"))
```
