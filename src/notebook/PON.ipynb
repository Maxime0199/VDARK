## *K*-mer filtering  
- **Intra-échantillon** (-ci2 sur chaque KMC DB avant union) → élimine le bruit de séquençage de chaque normal individuellement.
- **Inter-échantillon** (optionnel, si tu veux être plus conservateur) → au lieu d'une union simple, tu pourrais exiger qu'un k-mer soit présent dans ≥2 des N normaux du panel pour être retenu dans le panel final, plutôt que de garder tout ce qui apparaît dans ne serait-ce qu'un seul normal. Ça se fait avec kmc_tools en remplaçant l'union simple par une opération de type intersection/comptage seuils, ou plus simplement en post-traitant le dump.
